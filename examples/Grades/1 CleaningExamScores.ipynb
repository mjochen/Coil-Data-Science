# Interpreting exam results

We have a course where the exam consists of 2 parts: theory and practice. Theory is 25% of the final grade, practice 75%. There are seven groups of students divided over international students and local students.

## Data cleaning exam results

Let's begin by importing the data and checking it out.

```python
import pandas as pd
import re
```

Load the CSV and print the dataframe.

```python
#load in the grades files into a Pandas Dataframe
df_grades = pd.read_csv("grades_tot.csv",delimiter=";")

#Print out the entire dataframe
df_grades.head()
```

```python
df_grades.tail()
```

Remove the first and last rows as these don't contain usefull information.

```python
#The first row (Manual posting) and last row Test Student have no information, so we are dropping those rows
df_grades = df_grades.iloc[1:-1]

df_grades.head()
```

Print a list of all columns in the CSV.

```python
df_grades.columns
```

Select only the columns we can use.

```python
#Select the cols you want to inspect
name_id = 'SIS User ID' #OR df_grades.columns[2]
name_class = 'Section'  #OR df_grades.columns[4]
name_exam1 = 'Part 1: Multiple choice questions - January 2024 (Remotely Proctored) (473819)'
name_exam2 = 'Part 2: Practice - January 2024 _ CORRECT VERSION(Remotely Proctored) (475074)'
df_exam = df_grades[[name_id, name_class, name_exam1, name_exam2]]

df_exam.head()
```

Change the names of the columns into something more useable.

```python
#cleaning our column names:

df_exam.rename(columns={ 
name_id: 'id', 
name_class: 'class',
name_exam1: 'part1_theory',
name_exam2: "part2_exercise"
},inplace=True)


df_exam.head()
```

The first row contains the maximum score; we want to filter this row, but we will retain this useful information in the program (in a variable and in the column name).

```python

max_value_part1 = int(float(df_exam.loc[1, 'part1_theory']))
max_value_part2 = int(float(df_exam.loc[1, 'part2_exercise']))

print(f'the max value for part 1 = {max_value_part1}')
print(f'the max value for part 2 = {max_value_part2}')
```

```python
#cleaning our column names:

df_exam.rename(columns={ 
'part1_theory': 'part1_theory_'+str(max_value_part1),
'part2_exercise' : "part2_exercise_"+str(max_value_part2)
}, inplace=True)

#remove first row
df_exam = df_exam.iloc[1:].copy()

df_exam
```

Rename the classes to just the useful part using Regex.

```python
# Regex pattern
pattern = r'\b(ACS |TI |EI )\b(.{6})'

# Define a function to apply the regex and return the matched string or the "error message" Nope
def apply_regex(s):
    match = re.search(pattern, s)
    return match.group(2) if match else "Nope"

# Apply the function to the 'class' column
df_exam['class'] = df_exam['class'].copy().apply(apply_regex)

#print
df_exam
```

Reset the index so we start from 0 again.

```python
# Reset the index
df_exam = df_exam.reset_index(drop=True)

df_exam
```

And write this new dataframe to a CSV-file.

```python
# Save DataFrame to CSV
df_exam.to_csv('grades_clean.csv', index=False)
```
