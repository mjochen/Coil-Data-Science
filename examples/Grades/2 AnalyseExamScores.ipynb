# Interpreting the exam results

With clean data we can start making an analysis. What can we learn from this data?

## Starting from the clean CSV-file

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Read the data, make the s-number the index.

```python
#load in the grades files into a Pandas Dataframe
df_grades = pd.read_csv("grades_clean.csv", index_col='id')
df_input = pd.read_csv("grades_clean.csv", index_col='id')

#Print out the entire dataframe
df_grades.head(10)
```

Use .info() to understand the data.

```python
#Getting info about the data
print(f'the shape of the df = {df_grades.shape} \n')

print("Data information:")
df_grades.info()
```

How many NaN-values do we have?

```python
df_grades.isnull().sum()
```

Let's remove them!

```python
#Remove all student that didn't attend the exam
df_grades.dropna(inplace=True)
print(df_grades.shape)
df_grades.head(10)
```

Use .describe() to describe the data.

```python
df_grades.describe()
```

Classname is categorical data, so let's use it as such.

```python
df_grades["class"] = pd.Categorical(df_grades["class"])
df_grades['class'].unique()
```

Apply .describe() to every group separately.

```python
df_grades.groupby('class').describe()
```

Create a boxplot and a histogram for all results of part 1 + part 2.

```python
# Create a box plot for 'part1_theory_(30)'
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.boxplot(df_grades['part1_theory_30'])
plt.grid(True)
plt.title('Box plot of part1_theory_30')

plt.subplot(1, 2, 2)
plt.hist(df_grades['part1_theory_30'], bins=15, edgecolor='black')
plt.title('Histogram of part1_theory_30')
plt.grid(True)
plt.show()

# Create a box plot for 'part2_exercise_(10)'
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.boxplot(df_grades['part2_exercise_10'])
plt.title('Box plot of part2_exercise_10')
plt.grid(True)
plt.subplot(1, 2, 2)
plt.hist(df_grades['part2_exercise_10'], bins=10, edgecolor='black')
plt.title('Histogram of part2_exercise_10')
plt.grid(True)
plt.show()
```

Create a correlation matrix between the two parts.

```python
df_points = df_grades[["part1_theory_30", "part2_exercise_10"]]
correlation_matrix = df_points.corr()

# Plot the correlation matrix as a heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()
```

Create a scatter plot between columns A and B.

```python
# Create a scatter plot between columns A and B
plt.scatter(df_grades['part1_theory_30'], df_grades['part2_exercise_10'])
plt.xlabel('Part 1')
plt.ylabel('Part 2')
plt.title('Scatter Plot between part1 and part2')

# Fit a linear regression line
sns.regplot(x='part1_theory_30', y='part2_exercise_10', data=df_points, scatter=False, color='green')

plt.axhline(y=5, color='orange', linestyle='--', label='Y = 5')
plt.axvline(x=15, color='orange', linestyle='--', label='X = 15')
plt.grid(True)
plt.ylim(0, 10)
plt.yticks(np.arange(0, 11, 1))  # Set Y-axis ticks at intervals of 1


# Add text labels
plt.text(15.5, 8.5, 'Theo + Exer', color='green', fontsize=20)
plt.text(1.5, 4, 'Theo + Exer', color='red', fontsize=20)

plt.text(1.5, 8, 'Theo + ', color='red', fontsize=20)
plt.text(7.5, 8, 'Exer', color='green', fontsize=20)

plt.text(20, 4, 'Theo + ', color='green', fontsize=20)
plt.text(26.5, 4, 'Exer', color='red', fontsize=20)

plt.show()
```
